**Step 27: Detect Outliers using IQR**

```python
numeric_cols = ['Quantity', 'Unit_Price', 'Discount_Percent', 'Discount_Amount', 'Sales', 'Cost', 'Profit']

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    print(f"{col}: {len(outliers)} outliers (bounds: {lower_bound:.2f} to {upper_bound:.2f})")

def remove_outliers_iqr(data, cols):
    data_clean = data.copy()
    for col in cols:
        Q1 = data_clean[col].quantile(0.25)
        Q3 = data_clean[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        data_clean = data_clean[(data_clean[col] >= lower_bound) & (data_clean[col] <= upper_bound)]
    return data_clean

df_clean = remove_outliers_iqr(df, numeric_cols)
print("Before:", df.shape, "| After:", df_clean.shape)
```

**Step 28: Feature Engineering**

```python
df_fe = df_clean.copy()

df_fe['Order_Date'] = pd.to_datetime(df_fe['Order_Date'])
df_fe['Ship_Date'] = pd.to_datetime(df_fe['Ship_Date'])

df_fe['Order_Year'] = df_fe['Order_Date'].dt.year
df_fe['Order_Month'] = df_fe['Order_Date'].dt.month
df_fe['Order_Day'] = df_fe['Order_Date'].dt.day
df_fe['Order_Weekday'] = df_fe['Order_Date'].dt.dayofweek

df_fe['Shipping_Days'] = (df_fe['Ship_Date'] - df_fe['Order_Date']).dt.days
df_fe['Profit_Margin'] = df_fe['Profit'] / df_fe['Sales']
df_fe['Revenue_Per_Unit'] = df_fe['Sales'] / df_fe['Quantity']

df_fe.head()
```

**Step 29: Feature Selection**

```python
df_model = df_fe.drop(columns=['Order_ID', 'Order_Date', 'Ship_Date', 'Customer_Name'])

target = 'Profit'
X = df_model.drop(columns=[target])
y = df_model[target]

print("Features:", X.columns.tolist())
```

  **Step 30: Encoding Categorical Data**

```python
categorical_cols = ['Segment', 'Region', 'State', 'Category', 'Product', 'Ship_Mode', 'Payment_Mode']

X_encoded = pd.get_dummies(X, columns=categorical_cols, drop_first=True)
print("Shape after encoding:", X_encoded.shape)
```

**Step 31: Feature Scaling**

```python
from sklearn.preprocessing import StandardScaler

numeric_features = ['Quantity', 'Unit_Price', 'Discount_Percent', 'Discount_Amount',
                     'Sales', 'Cost', 'Order_Year', 'Order_Month', 'Order_Day',
                     'Order_Weekday', 'Shipping_Days', 'Profit_Margin', 'Revenue_Per_Unit']

scaler = StandardScaler()
X_encoded[numeric_features] = scaler.fit_transform(X_encoded[numeric_features])
```

**Step 32: Split Dataset**

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size=0.2, random_state=42)
print("Train:", X_train.shape, "| Test:", X_test.shape)
```

**Step 33: Choose Algorithm**

```python
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor(n_estimators=200, max_depth=10, random_state=42)
```

**Step 34: Train Model**

```python
model.fit(X_train, y_train)
print("Training complete.")
```

**Step 35: Evaluate Model**

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

y_pred = model.predict(X_test)

print("MAE:", mean_absolute_error(y_test, y_pred))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred)))
print("R² Score:", r2_score(y_test, y_pred))

plt.figure(figsize=(7,5))
plt.scatter(y_test, y_pred, alpha=0.5)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--')
plt.xlabel("Actual Profit")
plt.ylabel("Predicted Profit")
plt.title("Actual vs Predicted Profit")
plt.show()

importances = pd.Series(model.feature_importances_, index=X_encoded.columns).sort_values(ascending=False)
plt.figure(figsize=(8,6))
importances.head(15).plot(kind='barh')
plt.title("Top 15 Feature Importances")
plt.gca().invert_yaxis()
plt.show()
```

**Step 36: Save the Model**

```python
import joblib

joblib.dump(model, "sales_profit_model.pkl")
joblib.dump(scaler, "scaler.pkl")
joblib.dump(X_encoded.columns.tolist(), "model_columns.pkl")
print("Saved successfully.")
```
